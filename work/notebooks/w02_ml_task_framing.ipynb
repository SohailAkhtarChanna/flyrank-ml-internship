## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Check how common the target class is
target_rate = df["target"].mean()

print(f"Observed decline rate: {target_rate:.3f}")
print(f"Observed decline percentage: {target_rate * 100:.1f}%")
```

### Success metric

I would use **Precision@50** as the main success metric. The goal is to identify a small group of pages that are most likely to decline, so precision matters because the content team has limited time for reviews. A higher Precision@50 means that more of the top 50 prioritized pages were actually observed as declining.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Inspect several signals that could be useful for the task
print(df.columns)
cols = ["clicks_90d", "impressions_90d", "ctr", "avg_position", "trend_direction"]

df[cols].head(10)
```

### Why ML beats a fixed rule here

The pattern is too messy for a single fixed rule because page performance can depend on multiple signals at the same time, such as impressions, clicks, CTR, average position, and content characteristics. A rule such as "review every page below a certain position" would ignore these interactions and could prioritize many pages unnecessarily. ML can combine several measured signals to rank or classify pages based on their observed patterns. The output should be treated as **decision support**, not as a guarantee that a page will decline.
